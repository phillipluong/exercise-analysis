## 1.1 Clean the datasets
Turning the datasets into Pandas so I can translate the data analysis over easily

```python
# df = df.toPandas()

def safe_literal_eval(x):
    if pd.isna(x):
        return x
    try:
        result = ast.literal_eval(x)
        return set(result) if isinstance(result, (list, set)) else result
    except (ValueError, SyntaxError):
        return {x} if isinstance(x, str) else x

df['level'] = df['level'].apply(safe_literal_eval)
df['goal'] = df['goal'].apply(safe_literal_eval)
```

```python
# ex = ex.toPandas()

ex['level'] = ex['level'].apply(safe_literal_eval)
ex['goal'] = ex['goal'].apply(safe_literal_eval)
```
