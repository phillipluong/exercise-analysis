```python
cat2_conversion = {'Abdominals': 'Abs/Core', 'Adductors': 'Legs', 'Abductors': 'Legs', 'Biceps': 'Arms', \
'Calves': 'Legs', 'Chest': 'Chest', 'Forearms': 'Arms', 'Glutes': 'Legs', 'Hamstrings': 'Legs', 'Lats': 'Back', \
'Lower Back': 'Back', 'Middle Back': 'Back', 'Traps': 'Back', 'Neck': 'Shoulders/Neck', 'Quadriceps': 'Legs', \
'Shoulders': 'Shoulders/Neck', 'Triceps': 'Arms'}
```

```python
cat2 = np.array([])
for row in df2.itertuples():
    if row.Type in ['Stretching', 'Cardio']:
        cat2 = np.append(cat2, [row.Type])
    else:
        cat2 = np.append(cat2, [cat2_conversion[row.BodyPart]])
```

```python
df2['label'] = cat2
```

```python
df2
```

```python
df2_new = df2[['Title', 'label']]
```

```python
df2_new.rename(columns={'Title':'exercise'},inplace=True)
```
